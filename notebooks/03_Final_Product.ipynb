## **Description of the Final Infographic**

This infographic visualizes how NBA scoring evolved from 2010 to 2024 and highlights the strategic changes that reshaped the league’s offensive. It combines four data visualizations to communicate a clear narrative:

1. **Average Points per Game by Season**, showing the steep scoring rise beginning in 2014.
2. **Average 3-Point Attempts per Game by Season**, illustrating the league-wide adoption of perimeter shooting.
3. **A Scatter Plot of 3PA vs PTS**, showing a positive relationship (r = 0.427) between 3-point volume and scoring.
4. **A Histogram of Points by Era**, revealing the upward shift in the scoring distribution.

Together, these plots show how a combination of increased pace, improved spacing, and a growing emphasis on the 3-point shot reshaped the NBA’s scoring environment. While 3-pointers played a major role, the overall rise in scoring reflects broader changes in offensive strategy and efficiency across the league.
